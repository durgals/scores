```python
def energy_score(forecasts, obs):

    # must have dimensions of(num_events, num_ensembles)

    num_samples = forecasts.shape[1]

    s1 = 0.

    for i in range(num_samples):
        s1 += np.sqrt(np.sum(np.square(forecasts[:, i] - obs)))

    s2 = 0.
    for i in range(num_samples):
        for j in range(i, num_samples):
            s2 += 2 * np.sqrt(np.sum(np.square((forecasts[:,i] - forecasts[:,j]))))


    es = (s1 / num_samples) - s2 / (2 * pow(num_samples, 2))

    return es
```

```python
def crps_ecdf(f_ens, o):

    f_ens = np.sort(f_ens)

    o_cdf = f_cdf = f_prev = crps = 0
    m = len(f_ens)

    for i in range(m):
        f_n = f_ens[i]
        if o_cdf == 0 and o < f_n:
            crps += (o - f_prev) * f_cdf ** 2
            crps += (f_n - o) * (f_cdf - 1) ** 2
            o_cdf = 1
        else:
            crps += ((f_n - f_prev) * (f_cdf - o_cdf) ** 2)

        f_cdf = (i+1.)/float(m) # ecdf
        f_prev = f_n

    # obs to the right of the forecast.
    if o_cdf == 0:
        crps += o - f_n

    return crps
```

```python
import time
import numpy as np

ens = np.random.normal(200,20,50)
obs = 152.
```

```python
ens.shape
```

```python
tic = time.time()
score1 = crps_ecdf(ens, obs)
toc = time.time()
time1 = toc-tic
```

```python
tic = time.time()
score2 = energy_score(np.array([ens]), np.array([obs]))
toc = time.time()
time2 = toc-tic
```

```python
print('CRPS ECDF:   ', score1, '  time:', time1)
print('Energy Score:', score2, '  time:', time2)
```
